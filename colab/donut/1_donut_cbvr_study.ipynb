# Contextual-band verified recovery: Donut reader study

**Qwen localization + Donut reading.** This is a hybrid system, not a Donut-only localizer. Both band answers come from the same frozen Donut DocVQA reader. Baseline answers also use Donut. No answer labels or attack targets enter localization or recovery.

Reuses **csev_full_inputs.zip**, already in Drive. No extra dataset transfer is needed for this scope. Saves to a separate `donut_bands_hybrid_v1` output directory; never changes Qwen results or CARC.

### Study scope
- Development: 75 clean receipts + 135 saved attacks (210 cases).
- Reserved evaluation: 45 collected clean receipts + 396 saved attacks (441 cases).
- **Saved attacks were generated against Qwen. These are transfer attacks when evaluated on Donut**, not Donut standard/EOT white-box robustness results.
- Eleven methods: no defense, JPEG, resize voting, selective multiview, semantic consistency, OCR-only, OCR evidence, fixed band, band answer, full/band agreement, and contextual-band verified recovery.
- Direct attacks: ten predetermined reserved receipts × two budgets (4,8/255) × two restarts × three targets (Donut without defense, band agreement, verified recovery): **120 runs**. Both defenses use Qwen localization.
- Frozen geometry inherited from Qwen development: full-width band, minimum height 12%, vertical margin 3%, maximum height 35%; expanded band adds 3%, capped at 45%. No automatic tuning on Donut evaluation.

This evaluates reader generalization using a shared localizer. It does not establish standalone Donut localization, public-dataset attacked robustness, or full Donut EOT coverage. The selected receipt cohort came from the Qwen study; it is not selected for Donut correctness. Report all-case results and clean-correct denominators for native attacks.

The original localization prompt is retained to match the Qwen study; its coordinate wording is interpreted using the same processed-pixel conversion. The notebook is statically checked locally, but the Donut GPU path still needs to execute on Colab.

Use an A100 runtime. Do not interrupt the running Qwen notebook to launch this one; run sequentially or on a separate allocated runtime. Checkpoints persist in Drive; Colab runtime limits still apply.

## 2. Extract inputs onto the runtime's local disk
Drive holds the archive and outputs; model inference reads local image copies for speed. The full hash/budget check comes after environment setup. No existing dataset is deleted.


````python
def file_sha(path):
    h=hashlib.sha256()
    with Path(path).open('rb') as f:
        for block in iter(lambda:f.read(8*1024*1024),b''):h.update(block)
    return h.hexdigest()

if not (ROOT/'EXTRACTED.json').exists():
    assert not ROOT.exists(), 'Partial extraction exists. Choose a different ROOT path; do not erase results.'
    with zipfile.ZipFile(BUNDLE) as z:
        required=sum(i.file_size for i in z.infolist())
    assert shutil.disk_usage('/content').free > BUNDLE.stat().st_size+required+2*1024**3, 'Insufficient local disk for extraction.'
    temp=Path(tempfile.mkdtemp(prefix='csev_extract_',dir='/content'))
    print('Copying bundle from Drive...',flush=True)
    local=temp/'bundle.zip';shutil.copy2(BUNDLE,local)
    print('Extracting...',flush=True)
    with zipfile.ZipFile(local) as z:
        for item in z.infolist():
            assert (temp/item.filename).resolve().is_relative_to(temp)
        z.extractall(temp)
    local.unlink()
    source=temp/'source';source.mkdir()
    with tarfile.open(temp/'source.tar') as t:
        for m in t.getmembers():
            assert (source/m.name).resolve().is_relative_to(source)
            assert m.isfile() or m.isdir()
        t.extractall(source,filter='data')
    (temp/'EXTRACTED.json').write_text(json.dumps({'bundle_name':BUNDLE.name}))
    temp.rename(ROOT)
config=json.loads((ROOT/'csev_receipts_v1.json').read_text())
print('Source commit:',config['git_commit'])
print('Input files:',len(json.loads((ROOT/'file_hashes.json').read_text())))
````

## 4. Run helper and interrupt cleanup
All stage logs are written to Drive as they stream. If you interrupt a stage, the helper terminates its child process; rerun the stage to resume. Checkpoints should only be loaded from your own trusted study directory.


````python
def run_stage(stage,*extra):
    global process
    env=os.environ.copy();env['HF_HOME']=str(ROOT/'hf_cache')
    log_path=OUT/(stage+'_'+datetime.now().strftime('%Y%m%d_%H%M%S')+'.log')
    with log_path.open('w') as log:
        process=subprocess.Popen([str(PYTHON),'-u',str(RUNNER),stage,'--root',str(ROOT),'--out',str(OUT),*map(str,extra)],
                                 stdout=subprocess.PIPE,stderr=subprocess.STDOUT,text=True,env=env)
        try:
            for line in process.stdout:
                print(line,end='');log.write(line);log.flush()
            code=process.wait()
        except BaseException:
            process.terminate()
            try:process.wait(timeout=15)
            except subprocess.TimeoutExpired:process.kill();process.wait()
            raise
    if code:raise RuntimeError(f'{stage} failed (exit {code}). Log: {log_path}')
    print('Stage finished:',stage)
````

## 8. Direct Donut / hybrid adaptive attacks: 120 runs
No separate pilot. Enable after freezing. Ten reserved receipts × budgets 4 and 8/255 × two restarts × three attack targets: no defense, band agreement, and verified recovery. Each run permits 100 gradient steps and 100 verifier queries. Query cost includes several model calls.

The no-defense attack optimizes Donut's full-image answer; band agreement optimizes full-image and band answers; recovery optimizes both band answers. Donut uses exact official preprocessing in the forward pass with a differentiable BPDA approximation for gradients. Parity and finite-gradient checks run in the actual attack path.

For defended queries, Qwen re-localizes from the current attacked image. Localization is not differentiated; its boxes are held fixed within a gradient step. Additional query search explores discrete behavior. This is bounded adaptive testing, not an exhaustive attack or a robustness certificate. Each selected PNG is replayed against all three methods. Checkpoints resume individual runs, including unfinished gradient steps.
