# Defense and adaptive-attack analysis
Read-only comparison of clean utility, non-adaptive robustness, abstention, latency, and adaptive robustness.

```python
from pathlib import Path
import sys, pandas as pd, seaborn as sns, matplotlib.pyplot as plt
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT/'src'))
from result_analysis import load_table
records = load_table(ROOT/'outputs/defenses/defenses_validation_v1/merged/records.parquet')
```

```python
summary = records.groupby(['example_type','defense_id']).agg(trials=('experiment_key','size'), clean_accuracy=('clean_correct','mean'), target_asr=('adversarial_target_success','mean'), abstention=('abstained','mean'), agreement=('agreement','mean'), extra_passes=('additional_vlm_passes','mean'), runtime=('runtime_seconds','median')).reset_index()
summary
```

```python
nonadaptive = summary.query("example_type == 'non_adaptive_attack'")
fig, axes = plt.subplots(1,3,figsize=(15,4))
sns.barplot(nonadaptive, x='defense_id', y='target_asr', ax=axes[0]); axes[0].tick_params(axis='x', rotation=30); axes[0].set_title('Lower is better: non-adaptive ASR')
sns.barplot(nonadaptive, x='defense_id', y='abstention', ax=axes[1]); axes[1].tick_params(axis='x', rotation=30); axes[1].set_title('Abstention')
sns.barplot(nonadaptive, x='defense_id', y='extra_passes', ax=axes[2]); axes[2].tick_params(axis='x', rotation=30); axes[2].set_title('Compute overhead'); plt.tight_layout()
```
