# Transfer and digital robustness analysis
Read-only analysis of the terminal-generated transfer run.

```python
from pathlib import Path
import sys, pandas as pd, seaborn as sns, matplotlib.pyplot as plt
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT/'src'))
from result_analysis import load_table, rate_table
PATH = ROOT/'outputs/transfer_robustness/transfer_robustness_validation_v1/merged/records.parquet'
records = load_table(PATH)
```

```python
clean = records.query("example_type == 'clean' and status == 'complete'")
rate_table(clean, ['model_key','transform_id'], 'clean_target_correct')
```

```python
attacked = records.query("example_type == 'attack' and status == 'complete'")
table = attacked.groupby(['model_key','transform_id','epsilon_255']).adversarial_target_success.mean().reset_index()
for model in table.model_key.unique():
    pivot = table.query('model_key == @model').pivot(index='transform_id', columns='epsilon_255', values='adversarial_target_success')
    plt.figure(figsize=(8,6)); sns.heatmap(pivot, annot=True, fmt='.1%', vmin=0, vmax=1); plt.title(f'{model}: target survival'); plt.show()
```

```python
direct = attacked.query("transform_id == 'png_reload'")
rate_table(direct, ['model_key','dataset_name','epsilon_255','source_attack_success'], 'adversarial_target_success')
```
