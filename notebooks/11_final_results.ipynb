# Final project result tables
This notebook only reads completed merged outputs. It creates report-ready tables and figures without running any model.

```python
from pathlib import Path
import sys, pandas as pd
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT/'src'))
from result_analysis import load_table
paths = {
 'collected_baseline': ROOT/'outputs/baseline/baseline_collected_external_v1/merged/records.parquet',
 'collected_qwen': ROOT/'outputs/qwen_attack/qwen_collected_external_v1/merged/records.parquet',
 'collected_donut': ROOT/'outputs/donut_attack/donut_collected_external_v1/merged/records.parquet',
 'transfer': ROOT/'outputs/transfer_robustness/transfer_robustness_validation_v1/merged/records.parquet',
 'patch': ROOT/'outputs/eot_patch/eot_patch_collected_v1/merged/records.parquet',
 'defenses': ROOT/'outputs/defenses/defenses_validation_v1/merged/records.parquet'}
tables = {name: load_table(path) for name, path in paths.items()}
```

```python
external = pd.concat([tables['collected_qwen'].assign(model='Qwen'), tables['collected_donut'].assign(model='Donut')])
external.groupby(['model','epsilon_255']).agg(N=('record_id','size'), successes=('saved_image_success','sum'), ASR=('saved_image_success','mean'), median_LPIPS=('lpips','median'), median_SSIM=('ssim','median')).reset_index()
```

```python
tables['defenses'].groupby(['example_type','defense_id']).agg(N=('record_id','size'), target_ASR=('adversarial_target_success','mean'), clean_accuracy=('clean_correct','mean'), abstention=('abstained','mean'), extra_passes=('additional_vlm_passes','mean')).reset_index()
```

## Reporting rule
Always show denominators, clean accuracy, perceptibility, failures, abstention, and compute beside attack success. Keep validation, external-test, transfer, visible-patch, and adaptive-defense results in separate rows.
