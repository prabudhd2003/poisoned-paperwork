# Localized EOT patch analysis
Compares the no-EOT control with the EOT treatment. This visible-patch threat model is reported separately from full-page PGD.

```python
from pathlib import Path
import sys, pandas as pd, matplotlib.pyplot as plt
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT/'src'))
from result_analysis import load_table
records = load_table(ROOT/'outputs/eot_patch/eot_patch_collected_v1/merged/records.parquet')
records.groupby('control').agg(trials=('record_id','size'), targeted_asr=('adversarial_target_success','mean'), digital_survival=('digital_survival_fraction','mean'), runtime=('runtime_seconds','median'))
```

```python
records.boxplot(column='digital_survival_fraction', by='control', figsize=(7,5)); plt.suptitle(''); plt.title('Digital survival: control vs EOT'); plt.ylabel('Fraction of transforms preserving target');
```

```python
from PIL import Image
row = records.sort_values(['adversarial_target_success','digital_survival_fraction'], ascending=False).iloc[0]
display(Image.open(row.image_path)); print(row[['record_id','control','patch_box','adversarial_target_success','digital_survival_fraction']])
```
